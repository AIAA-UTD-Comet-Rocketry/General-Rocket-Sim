```python
import pandas as pd
import matplotlib.pyplot as plt

flightData = pd.read_csv("ReferencedFiles/flight_computer_data.csv", header=0)

startRowNumber = flightData.index[flightData["Velocity_Up"] > 0][0]
endRowNumber = flightData.index[flightData["Burnout_Coast"] == 1][0]

print(f"Start row number is; {startRowNumber}")
print(f"End row number is; {endRowNumber}")

remainingDf = flightData[startRowNumber:endRowNumber]

remainingDf = remainingDf.loc[:, ["Flight_Time_(s)"]]

remainingDf.to_csv("ReferencedFiles/requiredSampleTimes.csv", index=None)

massOverTime = pd.read_csv("ReferencedFiles/massMapping.csv", header = 0)
plt.plot(massOverTime["time"], massOverTime["mass"])
plt.xlabel("Time From Flight Start")
plt.ylabel("Rocket Mass (kg)")
plt.show()
```
